```python
df = pd.read_csv('../data/GamingStudy_data.csv', encoding="ISO-8859-1")
df.head(3)
```

```python
df.info()

# I will be pulling S.No, Timestamp, GAD1-7, Game, Platform, Hours, whyplay (rename), Gender, 
# Age, Work, Degree, Residence, Playstyle, GAD_T, and Residence_ISO3 for this analysis.
```

## Fix Timestamp datatype

```python
df['Date'] = pd.to_datetime(df['Timestamp'], unit='D', origin='1899-12-30').dt.normalize()
df.head()
```

## Create custom dataframe for analysis

```python
gaming_GAD_df = df[['S. No.', 'Date', 'GAD1', 'GAD2', 'GAD3', 'GAD4', 'GAD5', 'GAD6', 'GAD7', 'GAD_T', 'Game', 'Platform', 'Hours', 'whyplay', 'Gender', 'Age', 'Work', 'Degree', 'Residence', 'Residence_ISO3', 'Playstyle']].copy()

gaming_GAD_df.head()
```

```python
gaming_GAD_df.info()
```

## Investigate null rows

```python
gaming_GAD_df.isna().sum()
```

```python
gaming_GAD_df.shape
```

### Drop 30 na Hours rows to keep int datatype

```python
gaming_GAD_df = gaming_GAD_df.dropna(subset=['Hours'])
gaming_GAD_df.shape
```

### Fill na Degree and Work with 'unknown'

```python
gaming_GAD_df[['Degree', 'Work']] = gaming_GAD_df[['Degree', 'Work']].fillna('Unknown')
gaming_GAD_df.isna().sum()
```

### Investigate Residence_ISO3 na values 

```python
gaming_GAD_df[gaming_GAD_df[['Residence_ISO3']].isna().any(axis=1)].head()

# na is returned when residence is unknown.
```

#### Remove unknown rows in 'Residence'

```python
gaming_GAD_df = gaming_GAD_df.dropna(subset='Residence_ISO3')
gaming_GAD_df.shape
```

```python
gaming_GAD_df.isna().sum()
```

## Investigate outliers in 'Hours'

There are 168 hours in a week. I will be filtering out any values that exceed this number and note the high outliers as they did not account for sleep. The same will apply to the non-game hours column.

I will also remove any rows where 'Hours' + 'Non-game-hours' > 168.

Rows removed: 7

```python
gaming_GAD_df = gaming_GAD_df[gaming_GAD_df['Hours'] <=168]
```

```python
# Check highest hour rows
sorted_hours_df = gaming_GAD_df.sort_values(by='Hours', ascending=False)
sorted_hours_df
```
