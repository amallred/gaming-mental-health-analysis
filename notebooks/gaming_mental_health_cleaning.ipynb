## Create custom dataframe for analysis

```python
gaming_GAD_df = df[['S. No.', 'Date', 'GAD1', 'GAD2', 'GAD3', 'GAD4', 'GAD5', 'GAD6', 'GAD7', 'GAD_T', 'Gender', 'Age', 'Work', 'Degree', 'Residence_ISO3', 'Game', 'Platform', 'Hours', 'whyplay', 'Playstyle']].copy()

gaming_GAD_df.head()
```

```python
gaming_GAD_df.info()
```

## Investigate null rows

```python
gaming_GAD_df.isna().sum()
```

### Drop 30 na Hours rows to keep int datatype

```python
gaming_GAD_df = gaming_GAD_df.dropna(subset=['Hours'])
gaming_GAD_df.shape
```

### Fill na Degree and Work with 'unknown'

```python
gaming_GAD_df[['Degree', 'Work']] = gaming_GAD_df[['Degree', 'Work']].fillna('Unknown')
gaming_GAD_df.isna().sum()
```

### Investigate Residence_ISO3 na values 

```python
gaming_GAD_df[gaming_GAD_df[['Residence_ISO3']].isna().any(axis=1)].head()

# na is returned when residence is unknown.
```

#### Remove 140 unknown rows in 'Residence'

```python
gaming_GAD_df = gaming_GAD_df.dropna(subset='Residence_ISO3')
gaming_GAD_df.shape
```

```python
gaming_GAD_df.isna().sum()
```

## Investigate outliers in 'Hours'

There are 168 hours in a week. I will be filtering out any values that exceed this number and note the high outliers as they did not account for sleep. The same will apply to the non-game hours column.

I will also remove any rows where 'Hours' + 'Non-game-hours' > 168.

Rows removed: 7

```python
gaming_GAD_df = gaming_GAD_df[gaming_GAD_df['Hours'] <=168]
```

```python
# Check highest hour rows
sorted_hours_df = gaming_GAD_df.sort_values(by='Hours', ascending=False)
sorted_hours_df
```

```python
gaming_GAD_df.info()
```

### Remove 'work' column
Other dataset does not have enough information for comparative analysis

```python
gaming_GAD_df = gaming_GAD_df.drop(columns='Work')
gaming_GAD_df.info()
```

## Rename columns to match other dataset

```python
gaming_GAD_df = gaming_GAD_df.rename(columns={
    "S. No." : "participant_id",
    "Date" : "year", 
    "GAD_T" : "GAD_total",
    "Age" : "age",
    "Residence_ISO3" : "residence",
    "Game" : "game",
    "Platform" : "platform",
    "Hours" : "hours",
    "whyplay" : "reason",
    "Playstyle" : "playstyle"
})

gaming_GAD_df.info()
```

## Calculate and add missing columns

### GAD_cat column calculation and relocation

```python
# categories from both NIHS codebook

total = gaming_GAD_df['GAD_total']
condlist = [
    total < 5,
    (total >= 5) & (total < 10),
    (total >= 10) & (total < 15),
    total >= 15
]
choicelist = [1, 2, 3, 4]

gaming_GAD_df['GAD_cat'] = np.select(condlist, choicelist, default=8)

gaming_GAD_df.head()

# Resources:
#   https://www.geeksforgeeks.org/python/numpy-select-function-python/
#   https://medium.com/@michalwesleymnach/the-complete-guide-to-create-columns-based-on-multiple-conditions-in-pandas-dataframes-eedf2c0392a6 
```

```python
col_to_move = gaming_GAD_df.pop('GAD_cat')
gaming_GAD_df.insert(10, 'GAD_cat', col_to_move)
gaming_GAD_df.head()
```

### education_id calculation and relocation

```python
# Identify each degree option found in the data

gaming_GAD_df['Degree'].unique().tolist()
```

```python
condlist_ed = [
    gaming_GAD_df['Degree'] == 'Bachelor\xa0(or equivalent)',
    gaming_GAD_df['Degree'] == 'High school diploma (or equivalent)',
    gaming_GAD_df['Degree'] == 'Ph.D., Psy. D., MD (or equivalent)',
    gaming_GAD_df['Degree'] == 'Master\xa0(or equivalent)',
    gaming_GAD_df['Degree'] == 'Unknown'
]
choicelist_ed = [8, 12, 13, 9, 99]

gaming_GAD_df['education_id'] = np.select(condlist_ed, choicelist_ed, default=99)

gaming_GAD_df.head()

# I used ChatGPT to learn how to run the condlist_ed as a boolean check rather than
# listing the values alone.
```

```python
# Move education_id column
col_to_move = gaming_GAD_df.pop('education_id')
gaming_GAD_df.insert(14, 'education_id', col_to_move)
gaming_GAD_df.head()
```

```python
# Identify each gender/sex option found in the data

gaming_GAD_df['Gender'].unique().tolist()
```

```python
condlist_sex = [
    gaming_GAD_df['Gender'] == 'Male',
    gaming_GAD_df['Gender'] == 'Female',
    gaming_GAD_df['Gender'] == 'Other'
]
choicelist_sex = [1, 2, 3]

gaming_GAD_df['sex_id'] = np.select(condlist_sex, choicelist_sex, default=9)

gaming_GAD_df.head()
```

```python
# Move sex_id column
col_to_move = gaming_GAD_df.pop('sex_id')
gaming_GAD_df.insert(12, 'sex_id', col_to_move)
gaming_GAD_df.head()
```

```python
# Move Gender and Degree  columns to end
col_to_move = gaming_GAD_df.pop('Gender')
gaming_GAD_df.insert(21, 'Gender', col_to_move)

col_to_move = gaming_GAD_df.pop('Degree')
gaming_GAD_df.insert(12, 'Degree', col_to_move)


gaming_GAD_df.head()
```
