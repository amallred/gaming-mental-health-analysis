I will also remove any rows where 'Hours' + 'Non-game-hours' > 168.

```python
sorted_hrs_df = clean_df.sort_values(by='Hours', ascending=False)
sorted_hrs_df.head(15)

clean_df = clean_df[clean_df['Hours'] <= 168]
```
